```python
#202432075_NayshaSalsabilaPirman
# =====================
# IMPORT LIBRARY
# =====================
import re, random, pickle, os, copy, time
from collections import Counter
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import TensorDataset, DataLoader
from torch.nn.utils.rnn import pack_padded_sequence
import nltk
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score, precision_recall_fscore_support
from sklearn.utils.class_weight import compute_class_weight
import warnings
warnings.filterwarnings('ignore')

nltk.download('stopwords')

SEED = 42
random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)
torch.cuda.manual_seed_all(SEED)

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print('Device yang digunakan:', device)
os.makedirs('figures', exist_ok=True)
```

  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">2. PRA-PEMROSESAN DATA (PREPROCESSING)</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Memuat dataset, membersihkan data, melakukan normalisasi / tokenisasi sekuens, serta membagi data latih dan data uji.</p>

```python
#202432075_NayshaSalsabilaPirman
# =====================
# LOAD DATASET
# =====================
# Dataset: https://www.kaggle.com/datasets/arushchillar/disneyland-reviews
# Unggah DisneylandReviews.csv ke Colab (atau sesuaikan path dari repositori GitHub)
# encoding latin-1 diperlukan karena file mengandung karakter non-UTF-8
df = pd.read_csv('DisneylandReviews.csv', encoding='latin-1')

print('Dimensi dataset (baris, kolom):', df.shape)
print('\nNama kolom:', list(df.columns))
print('\nLima sampel data:')
print(df[['Review_Text', 'Rating', 'Branch']].head())
print('\nJumlah nilai kosong per kolom:')
print(df.isnull().sum())

# Membuang ulasan duplikat (teks yang sama persis)
n_awal = len(df)
df = df.drop_duplicates(subset='Review_Text').reset_index(drop=True)
print(f'\nDuplikat dibuang: {n_awal - len(df)} | Sisa data: {len(df)}')

# Membentuk label sentimen dari rating bintang
def rating_to_sentiment(r):
    if r <= 2:
        return 'negative'
    elif r == 3:
        return 'neutral'
    return 'positive'

df['sentiment'] = df['Rating'].apply(rating_to_sentiment)
print('\nDistribusi label sentimen:')
print(df['sentiment'].value_counts())
```

```python
#202432075_NayshaSalsabilaPirman
# =====================
# EKSPLORASI DATA
# =====================
colors = ['#FF6B6B', '#FFD166', '#06D6A0']   # negative, neutral, positive
sentiment_counts = df['sentiment'].value_counts()[['negative', 'neutral', 'positive']]

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(13, 5))
rating_counts = df['Rating'].value_counts().sort_index()
ax1.bar(rating_counts.index, rating_counts.values, color='#114F6A')
ax1.set_title('Distribusi Rating Ulasan', fontsize=14, fontweight='bold')
ax1.set_xlabel('Rating (bintang)')
ax1.set_ylabel('Jumlah Ulasan')
ax2.pie(sentiment_counts.values, labels=sentiment_counts.index, autopct='%1.1f%%', colors=colors, startangle=90)
ax2.set_title('Distribusi Sentimen Ulasan', fontsize=14, fontweight='bold')
ax2.axis('equal')
plt.tight_layout()
plt.savefig('figures/fig1_distribusi_sentimen.png', dpi=150, bbox_inches='tight')
plt.show()
print('Jumlah data per sentimen:')
print(sentiment_counts)

branch_ct = pd.crosstab(df['Branch'], df['sentiment'])[['negative', 'neutral', 'positive']]
fig, ax = plt.subplots(figsize=(10, 5))
branch_ct.plot(kind='bar', ax=ax, color=colors)
ax.set_title('Distribusi Sentimen per Cabang Disneyland', fontsize=14, fontweight='bold')
ax.set_xlabel('Cabang')
ax.set_ylabel('Jumlah Ulasan')
plt.xticks(rotation=15)
plt.tight_layout()
plt.savefig('figures/fig2_sentimen_per_cabang.png', dpi=150, bbox_inches='tight')
plt.show()
```

```python
#202432075_NayshaSalsabilaPirman
# =====================
# PRA-PEMROSESAN TEKS
# =====================
stemmer = PorterStemmer()
# Kata negasi dipertahankan karena membalik makna sentimen ("not good", "never again")
negations = {'no', 'not', 'nor', 'never', 'dont', 'didnt', 'doesnt', 'wasnt', 'isnt', 'cant', 'wont', 'couldnt', 'wouldnt'}
stop_words = set(stopwords.words('english')) - negations

def clean_text(text):
    text = str(text)
    text = re.sub(r'https?://[A-Za-z0-9./]+', '', text)
    text = re.sub(r'<.*?>', '', text)
    text = text.lower()
    # Pada sebagian ulasan kontraksi rusak ("wasn   t"); dipulihkan menjadi kata negasi 'not'
    text = re.sub(r'\b(?:didn|don|doesn|wasn|isn|couldn|wouldn|weren|aren|hadn|hasn|haven|shouldn|can|won)\s+t\b', ' not ', text)
    text = re.sub(r"[\'\u2019]", '', text)      # "don't" -> "dont"
    text = re.sub(r'\d+', '', text)
    text = re.sub(r'[^a-z\s]', ' ', text)
    text = re.sub(r'\s+', ' ', text).strip()
    return text

def preprocess_text(text):
    tokens = clean_text(text).split()
    tokens = [stemmer.stem(w) for w in tokens if w not in stop_words and len(w) > 2]
    return ' '.join(tokens)

df['cleaned_text'] = df['Review_Text'].apply(preprocess_text)
df = df[df['cleaned_text'].str.len() > 0].reset_index(drop=True)
print('Jumlah data setelah pembersihan:', len(df))

print('\nContoh teks sebelum dan sesudah pra-pemrosesan:')
for i in range(3):
    print(f"\nOriginal: {df['Review_Text'].iloc[i][:200]}")
    print(f"Cleaned : {df['cleaned_text'].iloc[i][:200]}")
    print('-' * 50)
```

```python
#202432075_NayshaSalsabilaPirman
# =====================
# SPLIT, VOCABULARY, PADDING, TENSOR
# =====================
label_names = ['negative', 'neutral', 'positive']
sentiment_mapping = {'negative': 0, 'neutral': 1, 'positive': 2}
texts_all = df['cleaned_text'].values
y_all = df['sentiment'].map(sentiment_mapping).values

X_train_txt, X_test_txt, y_train, y_test = train_test_split(texts_all, y_all, test_size=0.2, random_state=42, stratify=y_all)
X_train_txt, X_val_txt, y_train, y_val = train_test_split(X_train_txt, y_train, test_size=0.2, random_state=42, stratify=y_train)
print(f'Train: {len(X_train_txt)} | Validation: {len(X_val_txt)} | Test: {len(X_test_txt)}')

VOCAB_SIZE = 20000
MAX_LEN = 128
BATCH_SIZE = 64

word_counts = Counter(w for t in X_train_txt for w in t.split())
vocab = {'<PAD>': 0, '<OOV>': 1}
for word, _ in word_counts.most_common(VOCAB_SIZE - 2):
    vocab[word] = len(vocab)
vocab_size = len(vocab)
print('Ukuran vocabulary:', vocab_size)

def encode_text(text):
    ids = [vocab.get(w, 1) for w in text.split()][:MAX_LEN]
    if len(ids) == 0:
        ids = [1]
    length = len(ids)
    ids = ids + [0] * (MAX_LEN - length)
    return ids, length

def build_tensors(texts, labels):
    encoded = [encode_text(t) for t in texts]
    X = torch.tensor([e[0] for e in encoded], dtype=torch.long)
    L = torch.tensor([e[1] for e in encoded], dtype=torch.long)
    y = torch.tensor(labels, dtype=torch.long)
    return X, L, y

X_train, L_train, y_train_t = build_tensors(X_train_txt, y_train)
X_val, L_val, y_val_t = build_tensors(X_val_txt, y_val)
X_test, L_test, y_test_t = build_tensors(X_test_txt, y_test)
print('Bentuk X_train:', tuple(X_train.shape))

seq_lengths = [len(t.split()) for t in texts_all]
plt.figure(figsize=(9, 5))
plt.hist(seq_lengths, bins=60, range=(0, 500), color='skyblue', edgecolor='black')
plt.axvline(np.mean(seq_lengths), color='red', linestyle='--', label=f'Rata-rata: {np.mean(seq_lengths):.1f}')
plt.axvline(MAX_LEN, color='green', linestyle='--', label=f'MAX_LEN: {MAX_LEN}')
plt.title('Distribusi Panjang Sekuens', fontsize=14, fontweight='bold')
plt.xlabel('Panjang Sekuens (jumlah token)')
plt.ylabel('Frekuensi')
plt.legend()
plt.savefig('figures/fig3_panjang_sekuens.png', dpi=150, bbox_inches='tight')
plt.show()
print(f'Persentil ke-90 panjang sekuens: {np.percentile(seq_lengths, 90):.0f} token')

class_weights = compute_class_weight(class_weight='balanced', classes=np.unique(y_train), y=y_train)
class_weights_tensor = torch.tensor(class_weights, dtype=torch.float32)
print('Class weights:', dict(zip(label_names, np.round(class_weights, 4))))

train_loader = DataLoader(TensorDataset(X_train, L_train, y_train_t), batch_size=BATCH_SIZE, shuffle=True)
val_loader = DataLoader(TensorDataset(X_val, L_val, y_val_t), batch_size=BATCH_SIZE, shuffle=False)
test_loader = DataLoader(TensorDataset(X_test, L_test, y_test_t), batch_size=BATCH_SIZE, shuffle=False)
```

<div style="background-color: #114F6A; color: #FFFFFF; padding: 10px 18px; border-radius: 8px; border-left: 8px solid #1FA4B1; font-family: 'Segoe UI', sans-serif; margin-top: 16px;">
  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">3. PEMODELAN ARSITEKTUR JARINGAN (RNN / LSTM)</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Mendefinisikan kelas arsitektur model sekuensial menggunakan modul torch.nn sesuai algoritma yang dipilih.</p>
</div>

```python
#202432075_NayshaSalsabilaPirman
# =====================
# ARSITEKTUR MODEL LSTM
# =====================
class SequenceModel(nn.Module):
    def __init__(self, vocab_size, embed_dim, hidden_size, num_classes, dropout=0.3, pad_idx=0):
        super(SequenceModel, self).__init__()
        self.embedding = nn.Embedding(vocab_size, embed_dim, padding_idx=pad_idx)
        self.emb_dropout = nn.Dropout(dropout)
        self.lstm = nn.LSTM(embed_dim, hidden_size, batch_first=True)
        self.fc1 = nn.Linear(hidden_size, 64)
        self.relu = nn.ReLU()
        self.dropout = nn.Dropout(0.5)
        self.fc2 = nn.Linear(64, num_classes)

    def forward(self, x, lengths):
        emb = self.emb_dropout(self.embedding(x))
        packed = pack_padded_sequence(emb, lengths.cpu(), batch_first=True, enforce_sorted=False)
        _, (h_n, c_n) = self.lstm(packed)
        h = h_n[-1]
        out = self.fc2(self.dropout(self.relu(self.fc1(h))))
        return out
```

<div style="background-color: #114F6A; color: #FFFFFF; padding: 10px 18px; border-radius: 8px; border-left: 8px solid #1FA4B1; font-family: 'Segoe UI', sans-serif; margin-top: 16px;">
  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">4. INISIASI HYPERPARAMETER &amp; PROSES PELATIHAN (TRAINING)</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Menentukan konfigurasi hyperparameter (learning rate, epochs, loss function, optimizer) dan menjalankan siklus pelatihan model.</p>
</div>

```python
#202432075_NayshaSalsabilaPirman
# =====================
# HYPERPARAMETER & TRAINING
# =====================
epochs = 20
learning_rate = 0.001
embed_dim = 128
hidden_size = 128
patience = 4

model = SequenceModel(vocab_size, embed_dim, hidden_size, num_classes=3).to(device)
criterion = nn.CrossEntropyLoss(weight=class_weights_tensor.to(device))
optimizer = optim.Adam(model.parameters(), lr=learning_rate)
scheduler = optim.lr_scheduler.ReduceLROnPlateau(optimizer, mode='min', factor=0.2, patience=2, min_lr=0.0001)
print(model)
total_params = sum(p.numel() for p in model.parameters() if p.requires_grad)
print(f'Total parameter yang dilatih: {total_params:,}')

def evaluate(loader):
    model.eval()
    total_loss, correct, total = 0.0, 0, 0
    all_probs, all_labels = [], []
    with torch.no_grad():
        for xb, lb, yb in loader:
            xb, yb = xb.to(device), yb.to(device)
            logits = model(xb, lb)
            loss = criterion(logits, yb)
            total_loss += loss.item() * yb.size(0)
            probs = torch.softmax(logits, dim=1)
            correct += (probs.argmax(dim=1) == yb).sum().item()
            total += yb.size(0)
            all_probs.append(probs.cpu().numpy())
            all_labels.append(yb.cpu().numpy())
    return total_loss / total, correct / total, np.concatenate(all_probs), np.concatenate(all_labels)

train_losses, val_losses = [], []
train_accs, val_accs = [], []
best_val_loss, best_epoch, best_state, wait = float('inf'), 0, None, 0
start_time = time.time()

for epoch in range(1, epochs + 1):
    model.train()
    run_loss, correct, total = 0.0, 0, 0
    for xb, lb, yb in train_loader:
        xb, yb = xb.to(device), yb.to(device)
        optimizer.zero_grad()
        logits = model(xb, lb)
        loss = criterion(logits, yb)
        loss.backward()
        nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)
        optimizer.step()
        run_loss += loss.item() * yb.size(0)
        correct += (logits.argmax(dim=1) == yb).sum().item()
        total += yb.size(0)
    tr_loss, tr_acc = run_loss / total, correct / total
    va_loss, va_acc, _, _ = evaluate(val_loader)
    scheduler.step(va_loss)
    train_losses.append(tr_loss); train_accs.append(tr_acc)
    val_losses.append(va_loss); val_accs.append(va_acc)
    current_lr = optimizer.param_groups[0]['lr']
    print(f'Epoch {epoch:02d}/{epochs} | loss {tr_loss:.4f} | acc {tr_acc:.4f} | val_loss {va_loss:.4f} | val_acc {va_acc:.4f} | lr {current_lr:.5f}')
    if va_loss < best_val_loss:
        best_val_loss, best_epoch, wait = va_loss, epoch, 0
        best_state = copy.deepcopy(model.state_dict())
    else:
        wait += 1
        if wait >= patience:
            print(f'Early stopping pada epoch {epoch}. Model terbaik: epoch {best_epoch}')
            break

model.load_state_dict(best_state)
train_time = time.time() - start_time
print(f'Pelatihan selesai dalam {train_time:.1f} detik')
```

```python
#202432075_NayshaSalsabilaPirman
# =====================
# CONTOH PREDIKSI TEKS BARU
# =====================
def predict_sentiment(text):
    model.eval()
    ids, length = encode_text(preprocess_text(text))
    x = torch.tensor([ids], dtype=torch.long).to(device)
    l = torch.tensor([length], dtype=torch.long)
    with torch.no_grad():
        probs = torch.softmax(model(x, l), dim=1).cpu().numpy()[0]
    return label_names[int(probs.argmax())], float(probs.max()), probs

test_texts = [
    'Amazing day at Disneyland! The rides were fantastic and the staff were so friendly.',
    'Terrible experience. Queues were over two hours long and the food was overpriced.',
    'The park was okay, nothing special. Some rides were closed but others were fine.',
    'Worst trip ever. Rude staff, dirty park and the rides kept breaking down.',
    'Magical place for kids, we loved the fireworks and would definitely come back.'
]
rows = []
for t in test_texts:
    label, conf, probs = predict_sentiment(t)
    rows.append({'Teks': t[:45] + '...', 'Prediksi': label, 'Confidence': round(conf, 3), 'P(neg)': round(float(probs[0]), 3), 'P(neu)': round(float(probs[1]), 3), 'P(pos)': round(float(probs[2]), 3)})
pred_df = pd.DataFrame(rows)
print(pred_df.to_string(index=False))

pred_df.set_index('Teks')[['P(neg)', 'P(neu)', 'P(pos)']].plot(kind='barh', stacked=True, figsize=(10, 4), color=['#FF6B6B', '#FFD166', '#06D6A0'])
plt.title('Skor Probabilitas Sentimen pada Teks Baru', fontweight='bold')
plt.xlabel('Probabilitas')
plt.tight_layout()
plt.savefig('figures/fig7_contoh_prediksi.png', dpi=150, bbox_inches='tight')
plt.show()
```

```python
#202432075_NayshaSalsabilaPirman
# =====================
# EKSPOR HASIL & ANGKA UNTUK LAPORAN
# =====================
import json

p, r, f1, sup = precision_recall_fscore_support(test_labels, test_preds, labels=[0, 1, 2], zero_division=0)
macro_f1 = float(np.mean(f1))
weighted_f1 = float(np.average(f1, weights=sup))

torch.save(model.state_dict(), 'disneyland_sentiment_lstm.pt')
with open('vocab.pickle', 'wb') as fh:
    pickle.dump(vocab, fh)

hasil = {
    'epoch_dijalankan': len(train_losses), 'epoch_terbaik': best_epoch,
    'train_acc_terbaik': train_accs[best_epoch - 1], 'val_acc_terbaik': val_accs[best_epoch - 1],
    'train_loss_terbaik': train_losses[best_epoch - 1], 'val_loss_terbaik': val_losses[best_epoch - 1],
    'test_acc': test_acc, 'test_loss': test_loss, 'macro_f1': macro_f1, 'weighted_f1': weighted_f1,
    'durasi_detik': train_time, 'total_parameter': total_params, 'vocab_size': vocab_size,
    'per_kelas': {label_names[i]: {'precision': float(p[i]), 'recall': float(r[i]), 'f1': float(f1[i]), 'support': int(sup[i])} for i in range(3)},
    'confusion_matrix': cm.tolist()
}
with open('hasil_run.json', 'w') as fh:
    json.dump(hasil, fh, indent=2)

print('=' * 60)
print('ANGKA UNTUK LAPORAN (salin ke bagian berwarna kuning)')
print('=' * 60)
print(f"Epoch dijalankan / terbaik : {hasil['epoch_dijalankan']} / {hasil['epoch_terbaik']}")
print(f"Train acc / Val acc (terbaik): {hasil['train_acc_terbaik']:.4f} / {hasil['val_acc_terbaik']:.4f}")
print(f"Train loss / Val loss (terbaik): {hasil['train_loss_terbaik']:.4f} / {hasil['val_loss_terbaik']:.4f}")
print(f"Test acc / Test loss        : {test_acc:.4f} / {test_loss:.4f}")
print(f"Macro F1 / Weighted F1      : {macro_f1:.4f} / {weighted_f1:.4f}")
print(f"Total parameter / Vocab     : {total_params:,} / {vocab_size}")
print(f"Durasi pelatihan            : {train_time:.1f} detik")
print('\nPer kelas (precision, recall, f1, support):')
for i, n in enumerate(label_names):
    print(f'  {n:9s} {p[i]:.4f}  {r[i]:.4f}  {f1[i]:.4f}  {int(sup[i])}')
print('\nConfusion matrix (baris=aktual, kolom=prediksi):')
print(cm)
```
